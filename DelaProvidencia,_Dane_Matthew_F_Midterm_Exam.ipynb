```python
#modified version
def main():
 class TemperatureConversion:
  def __init__(self, temp=1):
   self._temp = temp

 class CelsiusToFahrenheit(TemperatureConversion):
  def conversion(self):
   return (self._temp * 9) / 5 + 32

 class CelsiusToKelvin(TemperatureConversion):
  def conversion(self):
   return self._temp + 273.15

 class FahrenheitToCelsius(TemperatureConversion):
  def conversion(self):
   return (self._temp - 32) * 5 / 9

 class KelvinToCelsius(TemperatureConversion):
  def conversion(self):
   return self._temp - 273.15

 tempInCelsius = float(input("Enter the temperature in Celsius: "))
 convert = CelsiusToKelvin(tempInCelsius)
 print(str(convert.conversion()) + " Kelvin")
 convert = CelsiusToFahrenheit(tempInCelsius)
 print(str(convert.conversion()) + " Fahrenheit")

 tempInFahrenheit = float(input("Enter the temperature in Fahrenheit: "))
 convert = FahrenheitToCelsius(tempInFahrenheit)
 print(str(convert.conversion()) + " Celsius")

 tempInKelvin = float(input("Enter the temperature in Kelvin: "))
 convert = KelvinToCelsius(tempInKelvin)
 print(str(convert.conversion()) + " Celsius")

main()
```

```python
import tkinter as tk
class changeColor:
  def __init__(self, root):
    self.root = root
    self.root.title("Special Midterm Exam in OOP")
    self.root.geometry("300x200")
    self.button = tk.Button(
        self.root, text = "Click to Change Color", command = self.change_color
    )
    def change_color(self):
      self.button.config(bg = "yellow")

    root = tk.Tk()
    app = changeColor(root)
    root.mainloop()
```

# Question3

```python
import tkinter as tk

class DaneMatthew:
    def __init__(self, root):
        self.root = root
        root.title("Midterm in OOP")
        root.geometry("500x300")

        tk.Label(root, text="Enter Your Fullname",
                 fg="red").place(x=30, y=100)

        self.name = tk.Entry(root)
        self.name.place(x=250, y=95)

        tk.Button(root, text="Click to display your Fullname",
                  fg="red", command=self.display).place(x=30, y=150)

        self.output = tk.Entry(root)
        self.output.place(x=250, y=145)

    def display(self):
        self.output.delete(0, tk.END)
        self.output.insert(0, self.name.get())


root = tk.Tk()
app = DaneMatthew(root)
root.mainloop()

```
